```python
# Parameters
raw_data = [{"id": 1, "name": "Alice", "department": "Engineering", "score": 85}, {"id": 2, "name": "Bob", "department": "Engineering", "score": 72}, {"id": 3, "name": "Charlie", "department": "Marketing", "score": 91}, {"id": 4, "name": "Diana", "department": "Engineering", "score": 64}, {"id": 5, "name": "Eva", "department": "Marketing", "score": 88}]
min_score = 80
result_path = "/opt/airflow/output/papermill_result.json"
```

```python
import json
from pathlib import Path

qualified = [record for record in raw_data if record.get('score', 0) >= int(min_score)]
processed_data = {
    'qualified_employees': qualified,
    'total_qualified': len(qualified),
    'average_score': sum(record['score'] for record in qualified) / len(qualified) if qualified else 0,
}
Path(result_path).parent.mkdir(parents=True, exist_ok=True)
Path(result_path).write_text(json.dumps(processed_data, indent=2), encoding='utf-8')
processed_data
```
